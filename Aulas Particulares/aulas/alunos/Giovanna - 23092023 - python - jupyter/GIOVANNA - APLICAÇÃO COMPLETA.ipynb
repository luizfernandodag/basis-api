# LENDO OS DADOS

```python
# ler arquivos csv


files = {'customers'    : 'DADOS\olist_customers_dataset.csv',
         'geolocation'  : 'DADOS\olist_geolocation_dataset.csv',
         'items'        : 'DADOS\olist_order_items_dataset.csv',
         'payment'      : 'DADOS\olist_order_payments_dataset.csv',
         'orders'       : 'DADOS\olist_orders_dataset.csv',
         'products'     : 'DADOS\olist_products_dataset.csv',
         'sellers'      : 'DADOS\olist_sellers_dataset.csv',
         'review'       : 'DADOS\olist_order_reviews_dataset.csv',
         }

dfs = {}
for key, value in files.items():
    dfs[key] = pd.read_csv(value)
    
dfs
```

# Cruzandos os dados (inner joins, etc)

```python
#Cruzamento gradativo

#customers ans geolocation
customers_location = dfs['customers'].merge(dfs['geolocation'], how='inner', left_on='customer_zip_code_prefix', right_on='geolocation_zip_code_prefix').drop_duplicates('customer_id', keep='first')

#pprint(customers_location)

#order + customer + geolocation
cusloc_order = customers_location.merge(dfs['orders'],\
                                        how = 'inner',\
                                        on = 'customer_id')

#pprint(cusloc_order)
# customer + geolocation + orders +item
cuslocord_item = cusloc_order.merge(dfs['items'],\
                                    how = "inner",\
                                    on = 'order_id'    
                                    )

#pprint(cuslocord_item)

cuslocordite_prod = cuslocord_item.merge(dfs['products'],\
                                        how='inner',\
                                        on='product_id')

#pprint(cuslocordite_prod)

cuslocordite_rev = cuslocordite_prod.merge(dfs['review'],\
                                           how = 'left',\
                                           on='order_id')

#pprint(cuslocordite_rev)


#Selecionando as colunas de interesse 


```
